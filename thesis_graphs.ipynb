# Experiments and Results chapter

```python
from analysis.clustering import kelly_colors
import matplotlib.pyplot as plt
import os
import json
from network import Model
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

# Figures 3.1-3.3

# Define model
model_dir='./../models/all_tasks_256_001_2'
model=Model(model_dir)
hp=model.hp

# Get data for the training
with open(os.path.join(model_dir,'log.json')) as file:

    data = json.load(file)
    x=data['trials']

    # Do perf vs trials plot
    fig1,ax1=plt.subplots(figsize=(10,5))

    for i, rule in enumerate(hp['rule_trains']): 
        ax1.plot(x,data[f'perf_{rule}'], c = kelly_colors[i+1], label=rule)

    ax1.legend(ncol=2, loc='lower right')
    ax1.set_ylabel('Perf')
    ax1.set_xlabel('Trials')
    ax1.set_title('Performance on the 20 tasks Vs number of trials')

    # Do cost vs trials plot
    fig2,ax2=plt.subplots(figsize=(10,5))

    for i, rule in enumerate(hp['rule_trains']): 
        ax2.plot(x,data[f'cost_{rule}'], c = kelly_colors[i+1], label=rule)

    ax2.legend(ncol=2, loc='upper right')
    ax2.set_ylabel('Cost')
    ax2.set_xlabel('Trials')
    ax2.set_title('Cost on the 20 tasks Vs number of trials')

    # Do perf vs trials plot limited to <90 perf
    fig3,ax3=plt.subplots(figsize=(10,5))

    for i, rule in enumerate(hp['rule_trains']): 

        y=data[f'perf_{rule}']
        y=[i if i<0.9 else None for i in y]

        ax3.plot(x,y, c = kelly_colors[i+1], label=rule)

    ax3.legend(ncol=2, loc='lower right')
    ax3.set_ylabel('Perf')
    ax3.set_xlabel('Trials')
    ax3.set_title('Performance on the 20 tasks Vs number of trials')
```
